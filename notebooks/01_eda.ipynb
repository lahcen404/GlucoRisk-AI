## 11. Unique Values and Value Counts

In this step, we examine the unique values and their frequencies in the main clinical variables.

This is especially important for detecting suspicious values such as `0`.

Although the dataset contains no explicit missing values (`NaN`), some clinical measurements may contain zero values that could represent missing or invalid measurements.

We will therefore count the number of zero values in each numerical clinical variable.

```python
clinical_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

(df[clinical_columns] == 0).sum()
```

## 12. Numerical Variables Distribution

In this step, we visualize the distribution of the numerical variables using histograms.

Histograms help us understand:

- how the values are distributed;
- where most observations are concentrated;
- whether a variable is symmetric or skewed;
- whether there are unusual values or possible outliers.

This visualization will help us better understand the data before preprocessing.

```python
df[clinical_columns].hist(figsize=(14, 10), bins=20)

plt.suptitle("Distribution of Clinical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

## 13. Outlier Detection

In this step, we investigate possible outliers in the numerical clinical variables.

An outlier is an observation that is unusually far from the majority of the data.

We will use boxplots to visually identify potential outliers.

Outliers are important because they can influence statistical analysis, clustering algorithms such as K-Means, and classification models.

```python
for column in clinical_columns:
    plt.figure(figsize=(6, 4))
    
    sns.boxplot(y=df[column])
    
    plt.title(f"Boxplot of {column}")
    plt.ylabel(column)
    plt.tight_layout()
    plt.show()
```
