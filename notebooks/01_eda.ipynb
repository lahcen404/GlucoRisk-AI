## 13. Outlier Detection

In this step, we investigate possible outliers in the numerical clinical variables.

An outlier is an observation that is unusually far from the majority of the data.

We will use boxplots to visually identify potential outliers.

Outliers are important because they can influence statistical analysis, clustering algorithms such as K-Means, and classification models.

```python
for column in clinical_columns:
    plt.figure(figsize=(6, 4))
    
    sns.boxplot(y=df[column])
    
    plt.title(f"Boxplot of {column}")
    plt.ylabel(column)
    plt.tight_layout()
    plt.show()
```

## 14. Correlation Analysis

In this step, we analyze the relationships between the numerical clinical variables.

Correlation measures how strongly two variables are related:

- A value close to **+1** means a strong positive relationship.
- A value close to **-1** means a strong negative relationship.
- A value close to **0** means little or no linear relationship.

The correlation matrix helps us identify which clinical variables are related to each other before applying clustering and classification.

```python
correlation_matrix = df[clinical_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Clinical Variables")
plt.tight_layout()
plt.show()
```

## Scaling Before KNN Imputation

KNN uses distances, so we standardize the features with `StandardScaler` to put them on a comparable scale.

- Mean ≈ 0
- Standard deviation ≈ 1
- `NaN` values remain unchanged

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(df[clinical_columns])
```

```python
print(X_scaled[:5])
```

##  KNN Imputation

KNN Imputer fills missing values by finding similar patients and using their values to estimate the missing data.

We use 5 nearest neighbors for the imputation.

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)

X_imputed_scaled = imputer.fit_transform(X_scaled)
```

##  Check KNN Imputation

We verify that the missing values have been successfully imputed.

```python
np.isnan(X_imputed_scaled).sum()
```

## Restore Original Feature Scale

After KNN imputation, the values are still standardized.

We use `inverse_transform()` to convert them back to their original units.

```python
X_imputed = scaler.inverse_transform(X_imputed_scaled)
```

```python
print(X_imputed[:5])
```

##  Create the Imputed DataFrame

We convert the imputed NumPy array back into a DataFrame and keep the original column names and indexes.

```python
df_imputed = pd.DataFrame(
    X_imputed,
    columns=clinical_columns,
    index=df.index
)

df_imputed.head()
```

## Distributions After KNN Imputation

We visualize the clinical variables after KNN imputation to check whether their distributions remain reasonable.

```python
df_imputed.hist(figsize=(14, 10), bins=20)

plt.suptitle("Distributions After KNN Imputation", fontsize=16)
plt.tight_layout()
plt.show()
```

##  Outlier Detection After KNN Imputation

We use boxplots again to check the variables after imputation and verify that KNN did not create unusual values.

```python
for column in clinical_columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(y=df_imputed[column])
    plt.title(f"Boxplot of {column} After KNN Imputation")
    plt.ylabel(column)
    plt.tight_layout()
    plt.show()
```
